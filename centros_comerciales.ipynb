# Análisis de ventas en centros comerciales con Python

## 1. Introducción

Este proyecto analiza un conjunto de datos de transacciones comerciales realizadas en distintos centros comerciales entre 2021 y 2023.

El objetivo es estudiar el comportamiento de las ventas, identificar los centros comerciales y las categorías de productos con mejores resultados y analizar la distribución temporal de la actividad comercial.

El análisis se desarrolla mediante Python y Pandas, utilizando operaciones de agrupación, agregación y tratamiento de fechas para obtener indicadores que faciliten la interpretación de los resultados.

### Objetivos del análisis

- Calcular los ingresos totales generados.
- Identificar los centros comerciales y las categorías con mayor actividad comercial.
- Analizar la evolución temporal de las ventas.
- Identificar las transacciones de mayor valor.
- Comparar los resultados comerciales entre establecimientos.

```python
import pandas as pd

df = pd.read_csv("data/datos_ventas_centros_comerciales.csv")

df.head()
```

## 2. Análisis de ingresos

### 2.1. Ingresos totales

Se calcula la facturación total del conjunto de centros comerciales, considerando el precio de los productos y las unidades vendidas en cada transacción.

```python
# Calcular el importe total de cada venta
df["total_venta"] = df["quantity"] * df["price"]
```

```python
# Calcular la facturación total
total_ingresos = df["total_venta"].sum()

print(f"Ingresos totales: {total_ingresos:,.2f}")
```

**Resultado:** Durante el periodo analizado, los centros comerciales registraron una facturación conjunta de aproximadamente 1.491 millones de liras turcas (TRY), que servirá como referencia para comparar el rendimiento entre centros y categorías de productos.

### 2.2. Centro comercial con mayor facturación

Se analiza la facturación acumulada por centro comercial para identificar el establecimiento que ha generado mayores ingresos durante el periodo estudiado.

```python
# Calcular la facturación por centro comercial
ventas_por_centro = df.groupby("shopping_mall")["total_venta"].sum()

# Identificar el centro con mayor facturación
centro_comercial_mas_ventas = ventas_por_centro.sort_values(ascending=False).index[0]

print(centro_comercial_mas_ventas)
```

**Resultado:** Mall of Istanbul fue el centro comercial con mayor facturación durante el periodo analizado.

### 2.3. Categoría con mayor volumen de ventas

Se analiza la cantidad de unidades vendidas por categoría para identificar qué tipo de producto concentra el mayor volumen de ventas.

```python
# Calcular las unidades vendidas por categoría
categorias = df.groupby("category")["quantity"].sum()

# Identificar la categoría con mayor volumen de ventas
categoria_mas_vendida = categorias.sort_values(ascending=False).index[0]

print(categoria_mas_vendida)
```

**Resultado:** La categoría de ropa (*Clothing*) registró el mayor número de unidades vendidas durante el periodo analizado.

### 2.4. Categoría del producto más caro

Se identifica la categoría a la que pertenece el producto con el precio unitario más alto registrado en el conjunto de datos.

```python
# Identificar la categoría del producto con el precio más alto
categoria_producto_mas_caro = df[df["price"] == df["price"].max()]["category"].values[0]

print(categoria_producto_mas_caro)
```

**Resultado:** El producto con el precio unitario más alto pertenece a la categoría de juguetes (*Toys*). Este resultado identifica la categoría del artículo más caro..

### 2.5. Factura más antigua registrada

Se identifica la primera factura registrada en el dataset a partir de la fecha de compra, con el fin de determinar el inicio de la actividad comercial analizada.

```python
# Convertir la columna al formato fecha
df["invoice_date"] = pd.to_datetime(df["invoice_date"], dayfirst=True)

# Identificar la factura más antigua
factura_mas_antigua = df.sort_values("invoice_date")["invoice_no"].values[0]

df[df["invoice_no"] == factura_mas_antigua][["invoice_no", "invoice_date"]]
```

**Resultado:** La factura `I192911` es la más antigua registrada en el conjunto de datos y permite identificar el inicio del periodo de ventas analizado.

### 2.6. Cantidad promedio de productos por transacción

Se calcula el número medio de unidades vendidas por transacción para conocer el volumen habitual de compra registrado en los centros comerciales.

```python
# Calcular la cantidad promedio de unidades por transacción
cantidad_promedio_por_transaccion = df.groupby("invoice_no")["quantity"].sum().mean()

print(f"Promedio de unidades por transacción: {cantidad_promedio_por_transaccion:.2f}")
```

**Resultado:** Durante el periodo analizado se vendieron una media de 3 unidades por transacción, lo que permite conocer el volumen medio de compra registrado en los centros comerciales.

### 2.7. Día con mayor número de transacciones

Se analiza el número de transacciones registradas diariamente para identificar la fecha con mayor actividad comercial durante el periodo estudiado.

```python
# Identificar el día con mayor número de transacciones
dia_con_mas_ventas = df.groupby("invoice_date").size().sort_values(ascending=False).index[0]

print(dia_con_mas_ventas)
```

**Resultado:** El 24 de noviembre de 2021 fue el día con mayor actividad comercial, con 159 transacciones registradas. Este dato refleja el volumen de operaciones, no necesariamente el día de mayor facturación.

### 2.8. Cliente con mayor gasto total

Se analiza el gasto acumulado por cliente para identificar al que realizó compras por un mayor importe durante el periodo estudiado.

```python
# Calcular el gasto total por cliente
gasto_por_cliente = df.groupby("customer_id")["total_venta"].sum()

# Identificar al cliente con mayor gasto
cliente_que_mas_gasto = gasto_por_cliente.sort_values(ascending=False).index[0]

print(cliente_que_mas_gasto)
```

**Resultado:** El cliente `C307063` registró el mayor gasto del periodo, con un importe total de 49.998 liras turcas (TRY).

### 2.9. Factura con mayor importe total

Se calcula el importe acumulado por factura para identificar la transacción de mayor valor registrada durante el periodo analizado.

```python
# Calcular el importe total por factura
importe_por_factura = df.groupby("invoice_no")["total_venta"].sum()

# Identificar la factura de mayor valor
factura_max_valor = importe_por_factura.sort_values(ascending=False).index[0]

print(factura_max_valor)
```

**Resultado:** La factura `I311640` registró el importe más alto del periodo analizado, con un total de 49.998 liras turcas (TRY).

### 2.10. Distribución porcentual de la facturación por categoría

Se calcula el porcentaje de facturación que representa cada categoría sobre el total de ingresos, con el objetivo de identificar cuáles tienen mayor peso en las ventas.

```python
# Calcular la facturación por categoría
ventas_categoria = df.groupby("category")["total_venta"].sum()

# Calcular la facturación total
total_ventas = ventas_categoria.sum()

# Calcular la distribución porcentual por categoría
distribucion_porcentual_ventas_por_categoria =((ventas_categoria / total_ventas) * 100).sort_values(ascending=False)

print(distribucion_porcentual_ventas_por_categoria)
```

**Resultado:** La categoría de ropa (*Clothing*) concentra el 34,50 % de la facturación total, seguida de cosmética (15,22 %) y alimentación y bebidas (14,92 %). Estas tres categorías representan conjuntamente el 64,64 % de los ingresos, lo que refleja una elevada concentración de las ventas en estos segmentos.

### 2.11. Día de la semana con mayor actividad por centro comercial

Se analiza la distribución semanal de las transacciones para identificar el día con mayor actividad comercial en cada establecimiento.

```python
# Obtener el día de la semana de cada transacción
df["dia_semana"] = df["invoice_date"].dt.day_name()

# Contar las transacciones por centro comercial y día
ventas_por_centro_y_dia = df.groupby(["shopping_mall", "dia_semana"]).size()

# Identificar el día con más transacciones en cada centro
dia_semana_mas_ventas_por_centro = ventas_por_centro_y_dia.groupby(level=0).idxmax().apply(lambda x: x[1])

print(dia_semana_mas_ventas_por_centro)
```

**Resultado:** El viernes, el lunes y el martes son los días con mayor número de transacciones en tres centros comerciales cada uno. El domingo destaca únicamente en Cevahir AVM. Estos resultados muestran diferencias en la actividad comercial semanal entre establecimientos.
